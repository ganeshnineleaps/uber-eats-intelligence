```python

# DAY 6 – END-TO-END ML PIPELINE & BUSINESS INSIGHTS

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

from IPython.display import display

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.2f}".format)

# --------------------------------------------------
# 1. LOAD EXISTING PROJECT RESULTS
# --------------------------------------------------

files = {
    "customers": "/home/ninelaps/uber-ml/data/csv/customer_segments_final.csv",
    "restaurants": "/home/ninelaps/uber-ml/data/csv/restaurant_features.csv",
    "sentiment": "/home/ninelaps/uber-ml/data/csv/sentiment_results.csv",
    "delivery": "/home/ninelaps/uber-ml/data/csv/delivery_predictions.csv"
}

data = {}

for name, file in files.items():
    if os.path.exists(file):
        data[name] = pd.read_csv(file)
        print(f"{name}: Loaded {data[name].shape}")
    else:
        print(f"{name}: File not found - {file}")

customer_df = data.get("customers")
restaurant_df = data.get("restaurants")
sentiment_df = data.get("sentiment")
delivery_df = data.get("delivery")

# Check available columns
for name, df in data.items():
    print(f"\n{name.upper()} COLUMNS:")
    print(df.columns.tolist())


# --------------------------------------------------
# 2. CUSTOMER SEGMENTATION INSIGHTS
# --------------------------------------------------

if customer_df is not None:

    print("\nCUSTOMER SEGMENTATION")
    display(customer_df.head())

    # Find a likely cluster column
    cluster_col = next(
        (c for c in customer_df.columns
         if "cluster" in c.lower()),
        None
    )

    if cluster_col:
        customer_summary = (
            customer_df.groupby(cluster_col)
            .size()
            .reset_index(name="customer_count")
        )

        customer_summary["customer_percentage"] = (
            customer_summary["customer_count"]
            / customer_summary["customer_count"].sum() * 100
        )

        print("\nCustomer segment distribution:")
        display(customer_summary)

        customer_summary.plot(
            x=cluster_col,
            y="customer_count",
            kind="bar",
            legend=False,
            title="Customer Segment Distribution"
        )
        plt.ylabel("Number of Customers")
        plt.tight_layout()
        plt.show()

    # Profile clusters using available numeric columns
    if cluster_col:
        numeric_cols = customer_df.select_dtypes(
            include=np.number
        ).columns.tolist()

        numeric_cols = [
            c for c in numeric_cols
            if c != cluster_col
        ]

        if numeric_cols:
            print("\nCustomer segment profiles:")
            display(
                customer_df.groupby(cluster_col)[numeric_cols]
                .mean()
                .round(2)
            )


# --------------------------------------------------
# 3. RESTAURANT SEGMENTATION INSIGHTS
# --------------------------------------------------

if restaurant_df is not None:

    print("\nRESTAURANT SEGMENTATION")
    display(restaurant_df.head())

    restaurant_cluster = next(
        (c for c in restaurant_df.columns
         if "cluster" in c.lower()),
        None
    )

    if restaurant_cluster:

        restaurant_summary = (
            restaurant_df.groupby(restaurant_cluster)
            .size()
            .reset_index(name="restaurant_count")
        )

        restaurant_summary["restaurant_percentage"] = (
            restaurant_summary["restaurant_count"]
            / restaurant_summary["restaurant_count"].sum() * 100
        )

        print("\nRestaurant segment distribution:")
        display(restaurant_summary)

        numeric_cols = restaurant_df.select_dtypes(
            include=np.number
        ).columns.tolist()

        numeric_cols = [
            c for c in numeric_cols
            if c != restaurant_cluster
        ]

        if numeric_cols:
            print("\nRestaurant segment profiles:")
            display(
                restaurant_df.groupby(restaurant_cluster)[numeric_cols]
                .mean()
                .round(2)
            )


# --------------------------------------------------
# 4. SENTIMENT ANALYSIS INSIGHTS
# --------------------------------------------------

if sentiment_df is not None:

    print("\nSENTIMENT ANALYSIS")
    display(sentiment_df.head())

    sentiment_col = next(
        (c for c in sentiment_df.columns
         if "sentiment" in c.lower()),
        None
    )

    if sentiment_col:

        sentiment_summary = (
            sentiment_df[sentiment_col]
            .value_counts()
            .reset_index()
        )

        sentiment_summary.columns = [
            "sentiment",
            "review_count"
        ]

        sentiment_summary["percentage"] = (
            sentiment_summary["review_count"]
            / sentiment_summary["review_count"].sum() * 100
        )

        print("\nSentiment distribution:")
        display(sentiment_summary)

        sentiment_summary.plot(
            x="sentiment",
            y="review_count",
            kind="bar",
            legend=False,
            title="Customer Sentiment Distribution"
        )
        plt.ylabel("Number of Reviews")
        plt.tight_layout()
        plt.show()

    # Inspect negative reviews and their text, if available
    review_text_col = next(
        (c for c in sentiment_df.columns
         if c.lower() in ["review_text", "cleaned_review", "text"]),
        None
    )

    if sentiment_col and review_text_col:
        negative_reviews = sentiment_df[
            sentiment_df[sentiment_col]
            .astype(str).str.lower().str.contains("negative")
        ]

        print("\nExamples of negative reviews:")
        display(negative_reviews[[sentiment_col, review_text_col]].head(10))


# --------------------------------------------------
# 5. DELIVERY PREDICTION INSIGHTS
# --------------------------------------------------

if delivery_df is not None:

    print("\nDELIVERY PREDICTION")
    display(delivery_df.head())

    # Identify actual and predicted delivery columns
    actual_col = next(
        (c for c in delivery_df.columns
         if c.lower() in [
             "actual_delivery_time",
             "actual_delivery_time_minutes",
             "delivery_time_minutes",
             "actual"
         ]),
        None
    )

    predicted_col = next(
        (c for c in delivery_df.columns
         if c.lower() in [
             "predicted_delivery_time",
             "predicted_delivery_time_minutes",
             "predicted_delivery_time_minutes_rf",
             "predicted"
         ]),
        None
    )

    if actual_col and predicted_col:

        delivery_df["prediction_error"] = (
            delivery_df[actual_col] - delivery_df[predicted_col]
        )

        delivery_df["absolute_error"] = (
            delivery_df["prediction_error"].abs()
        )

        mae = delivery_df["absolute_error"].mean()

        rmse = np.sqrt(
            (delivery_df["prediction_error"] ** 2).mean()
        )

        print("\nDelivery prediction evaluation")
        print("MAE:", round(mae, 2))
        print("RMSE:", round(rmse, 2))

        plt.figure(figsize=(8, 5))
        plt.scatter(
            delivery_df[actual_col],
            delivery_df[predicted_col],
            alpha=0.5
        )

        plt.xlabel("Actual Delivery Time")
        plt.ylabel("Predicted Delivery Time")
        plt.title("Actual vs Predicted Delivery Time")
        plt.tight_layout()
        plt.show()

    else:
        print(
            "Could not identify actual and predicted delivery columns. "
            "Check the printed column names."
        )


# --------------------------------------------------
# 6. DEMAND FORECASTING
# --------------------------------------------------

# If forecast_results was created in the same notebook/kernel,
# use it directly. Otherwise, load a forecast results file
# after exporting it from your forecasting notebook.

if "forecast_results" in globals():

    demand_df = forecast_results.copy()

    print("\nDEMAND FORECAST")
    display(demand_df.head())

    actual_demand_col = "actual_demand"
    predicted_demand_col = "predicted_demand"

    if (
        actual_demand_col in demand_df.columns
        and predicted_demand_col in demand_df.columns
    ):

        demand_mae = (
            demand_df[actual_demand_col]
            - demand_df[predicted_demand_col]
        ).abs().mean()

        demand_rmse = np.sqrt(
            (
                demand_df[actual_demand_col]
                - demand_df[predicted_demand_col]
            ) ** 2
        ).mean()

        print("Forecast MAE:", round(demand_mae, 2))
        print("Forecast RMSE:", round(demand_rmse, 2))

        plt.figure(figsize=(11, 5))

        plt.plot(
            demand_df["date"],
            demand_df[actual_demand_col],
            marker="o",
            label="Actual Demand"
        )

        plt.plot(
            demand_df["date"],
            demand_df[predicted_demand_col],
            marker="o",
            label="Predicted Demand"
        )

        plt.title("Actual vs Predicted Demand")
        plt.xlabel("Date")
        plt.ylabel("Demand")
        plt.legend()
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

    else:
        print("Check the column names in forecast_results.")

else:
    print(
        "\nDemand forecast not available in this kernel. "
        "Run the forecast_results creation cell from your "
        "updated forecasting notebook in this notebook first."
    )


# --------------------------------------------------
# 7. INTEGRATED BUSINESS INSIGHTS
# --------------------------------------------------

print("\n" + "=" * 60)
print("INTEGRATED BUSINESS INSIGHTS")
print("=" * 60)

print("""
1. CUSTOMER SEGMENTATION
   Identify high-value and frequent customers.
   Create retention campaigns for valuable or inactive segments.

2. RESTAURANT SEGMENTATION
   Identify restaurants with weak operational performance.
   Provide targeted support to improve preparation and delivery.

3. SENTIMENT ANALYSIS
   Understand the major reasons behind negative customer reviews.
   Prioritize improvements based on recurring complaints.

4. DELIVERY PREDICTION
   Use estimated delivery times to improve customer expectations.
   Identify orders that may require operational attention.

5. DEMAND FORECASTING
   Anticipate future demand and prepare delivery capacity.
   Allocate delivery partners according to expected demand.

6. OVERALL BUSINESS VALUE
   Combine customer, restaurant, review, delivery, and demand
   insights to improve customer satisfaction and operational efficiency.
""")


# --------------------------------------------------
# 8. LIMITATIONS
# --------------------------------------------------

print("\nMODEL LIMITATIONS")

print("""
- Historical data may not represent future customer behaviour.
- Clustering results depend on selected features and parameters.
- Sentiment models may misinterpret sarcasm and complex language.
- Delivery predictions can be affected by unexpected traffic,
  weather, and restaurant preparation delays.
- Demand forecasts may not capture sudden events or promotions.
- Model predictions should be monitored before operational use.
""")


# --------------------------------------------------
# 9. FUTURE IMPROVEMENTS
# --------------------------------------------------

print("\nFUTURE IMPROVEMENTS")

print("""
- Add weather, traffic, and promotional information.
- Improve model performance through hyperparameter tuning.
- Validate models across different locations and time periods.
- Monitor model drift and retrain models when necessary.
- Build an integrated dashboard for business teams.
- Test recommendations through controlled business experiments.
""")
```
